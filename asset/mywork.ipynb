```python
import pandas as pd
import json
```

```python
with open("t20_json_files/t20_json_files/t20_wc_match_results.json") as f:
    data=json.load(f)

Matches=pd.DataFrame(data[0]["matchSummary"])
Matches.head()
```

```python
Matches.rename({"scorecard":"match_id"},axis=1,inplace=True)
Matches.head()
```

```python
Matches.to_csv("Match_summary.csv",index=False)
```

```python
match_dict={}

for i,rows in Matches.iterrows():
    key1=rows["team1"]+" Vs "+rows["team2"]
    key2=rows["team2"]+" Vs "+rows["team1"]

    match_dict[key1]=rows["match_id"]
    match_dict[key2]=rows["match_id"]

```

Batting Summary


```python
with open("t20_json_files/t20_json_files/t20_wc_batting_summary.json") as f:
    data=json.load(f)
dat=[]
for rec in data:
    dat.extend(rec["battingSummary"])

Batting=pd.DataFrame(dat)
Batting.head()
```

```python
Batting["Out/Not"]=Batting.dismissal.apply(lambda x: "Out" if len(x)>0 else "Not Out")
Batting.drop("dismissal",axis=1,inplace=True)
```

```python
Batting
```

```python
Batting["match_id"]=Batting["match"].apply(lambda x:match_dict[x])
Batting
```

```python
Batting.to_csv("Batting_records_2022.csv",index=False)
```

Bowling Records


```python
with open("t20_json_files/t20_json_files/t20_wc_bowling_summary.json") as f:
    data=json.load(f)

bowl=[]
for rec in data:
    bowl.extend(rec["bowlingSummary"])

Bowling=pd.DataFrame(bowl)
Bowling.head()
```

```python
Bowling["match_id"]=Bowling["match"].map(match_dict)
Bowling
```

```python
Bowling.to_csv("Bowling_records.csv",index=False)
```
